```python
import pandas as pd
```

```python
corpus = [
    "Data science is fun",
    "NLP is a part of data science",
    "Students learn NLP with Python",
    "Python is useful for data science"
]

#query
query = "What is python used for in data science?"
```

```python
pip install rank_bm25
```

```python
from rank_bm25 import BM25Okapi

tokenized_corpus = [doc.lower().split() for doc in corpus]
tokenized_query = query.lower().split()
```

```python
#step2: get scores
bm25 = BM25Okapi(tokenized_corpus)
doc_scores = bm25.get_scores(tokenized_query)
```

```python
#step3: display results
for i, score in enumerate(doc_scores):
    print(f"Document {i+1}: Score = {score:.4f} | {corpus[i]}") 

#higher score means more relevant to the query.
```
